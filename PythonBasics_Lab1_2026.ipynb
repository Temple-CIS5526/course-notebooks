
# Typing

In addition to `float`, Python has several important built-in types that are worth knowing. One group consists of *simple types* that allow handling of numerical data, strings, Boolean values, dates, and time. Another group consists of *container types* that can contain numerous objects.

## Simple Types

**Numbers**. Integers and floats work as you would expect from other languages:


````python

c = 3
print(c, type(c))
````

````python

print(c + 1)    # Addition (by the way, any text following '#' is treated as a comment)
print(c * 2)    # Multiplication
print(c / 2)    # Division; in Python 3 the result is 1.5
print(c // 2)   # Integer (floor) division; the result is 1
print(c ** 2)   # Exponentiation
````

````python

c += 1
print(c)  # Prints "4"
c *= 2
print(c)  # Prints "8"
````


Operations between `float` and `integer` types are possible and the result is typed as `float`.


````python

print(b+c, type(b+c))
````


It is also possible to convert one type into another by using the built-in functions `float` and `int`.


````python

b = int(b)
print(b, type(b))
c = float(c)
print(c, type(c))
````


When we write `b = int(b)`, Python first computes `int(b)` and then assigns the resulting integer object to the name `b`. Python does not require us to declare a fixed type for the name `b`; the object that `b` currently refers to has a type. The same idea applies to the conversion of `c` to a float above.

In Python 3, integers are represented by the `int` type and can grow to very large values as needed. Python also has a built-in type for complex numbers. More details are available at https://docs.python.org/3/library/stdtypes.html#numeric-types-int-float-complex.

**None**. Another important simple value is `None`, which is used quite often in data science:


````python

n = None
print(n, type(n))
````


**Booleans**. Python implements the usual operators for Boolean logic using the readable words `and`, `or`, and `not`.

Python also has operators such as `&` and `|`, but they are not simply alternative spellings of `and` and `or`; they are bitwise operators and later also have special uses with packages such as NumPy and pandas.


````python

t = True
f = False
print(t, type(t), f, type(f))
````

````python

print(t and f)  # Logical AND
print(t or f)   # Logical OR
print(not t)    # Logical NOT
print(t != f)   # True when the two Boolean values are different
````


**Strings**. `str` is a very useful type. It is defined by assigning a *string literal* enclosed with single or double quotes.


````python

s1 = 'hello'       # string literals can use single quotes
print(s1, len(s1), type(s1))  # len() is a built-in function; check len? or help("len")
s2 = "hello"       # string literals can also use double quotes
print(s2)
s3 = '''
longer text
  that can span
 multiple lines
'''
print(s3)          # triple quotes can be used for multi-line string literals
s1 == s2           # this will output True because the objects have the same content
````

````python

s4 = s1 + ' is the same as ' + s2       # string concatenation
print(s4)

s5 = '%s %s %d' % (s1, 'world', 12)    # older printf-style string formatting; still valid
print(s5)                                # prints "hello world 12"

s6 = f'{s1} world {12}'                 # f-string: common modern Python string formatting
print(s6)                                # also prints "hello world 12"
````


String objects have many useful methods (see https://docs.python.org/3/library/stdtypes.html#string-methods for more details).


````python

s = "hello"
print(s.capitalize())             # Capitalize a string; prints "Hello"
print(s.upper())                  # Convert a string to uppercase; prints "HELLO"
print(s.rjust(7))                 # Right-justify a string, padding with spaces; prints "  hello"
print(s.center(7))                # Center a string, padding with spaces; prints " hello "
print(s.replace('l', '(ell)'))    # Replace all instances of one substring with another
                                 # prints "he(ell)(ell)o"
print('  world '.strip())         # Strip leading and trailing whitespace; prints "world"
````


## Container Types

Python includes several built-in container types: lists, dictionaries, sets, and tuples.

**Lists** (reference: https://docs.python.org/3/tutorial/datastructures.html#more-on-lists). A list is similar to an array in many languages, but it is resizable and can contain elements of different types. It is created by using square brackets.


````python

xs = [3, 1, 2]   # Create a list
print(xs)
````


Each element of the list can be accessed using bracket notation. Each element can be changed. The elements of the list do not have to be of the same type. Elements can be added or deleted using the appropriate methods.


````python

print(xs[0])             # the first element of a list has index 0
xs[1] = 'hello again'    # changing the second element from integer to string type
print(xs)
print(type(xs), type(xs[0]), type(xs[1]))
xs.append(99.99)         # add a new element at the end of the list
print(xs)
x = xs.pop(2)            # remove the third element and return its value
print(x, xs)
````


Lists can contain lists as elements.


````python

xbig = [xs, 1, 'hello']
print(xbig)
````


In addition to accessing list elements one at a time, Python provides concise syntax to access sublists; this is known as slicing.


````python

nums = list(range(5))     # range(5) represents 0,1,2,3,4; list(...) makes a modifiable list
print(nums)               # Prints "[0, 1, 2, 3, 4]"
print(nums[2:4])          # Get a slice from index 2 to 4 (exclusive); prints "[2, 3]"
print(nums[2:])           # Get a slice from index 2 to the end; prints "[2, 3, 4]"
print(nums[:2])           # Get a slice from the start to index 2 (exclusive); prints "[0, 1]"
print(nums[:])            # Get a slice of the whole list; prints "[0, 1, 2, 3, 4]"
print(nums[:-1])          # Slice indices can be negative; prints "[0, 1, 2, 3]"
nums[2:4] = [88, 99]     # Assign a new sublist to a slice
print(nums)               # Prints "[0, 1, 88, 99, 4]"
````


**Tuples** (reference: https://docs.python.org/3/tutorial/datastructures.html#tuples-and-sequences). A tuple is an immutable ordered sequence of values. It is usually written using parentheses.


````python

t = (5, 6)       # Create a tuple
print(t, type(t))
````


A tuple is in many ways similar to a list. One of the most important differences is that tuples can be used as keys in dictionaries (defined next) and as elements of sets, while lists cannot.


````python

dict2 = {(x, x + 1): x for x in range(6)}  # Create a dictionary with tuple keys (do not worry about the logic for now)
print(dict2)
print(dict2[t])
print(dict2[(1, 2)])
````

````python

t[0] = 1         # tuples are immutable, so this line intentionally produces a TypeError
````


**Dictionaries** (reference: https://docs.python.org/3/library/stdtypes.html#dict). A dictionary stores (key, value) pairs called *items*, similar to a Map in Java or an object in JavaScript. It is created using curly braces.


````python

dct = {'cat': 'cute', 'dog': 'furry'}   # Create a new dictionary with some data
print(dct, type(dct))
print(dct['cat'])                        # Get an entry from a dictionary; prints "cute"
print(list(dct.keys()))                  # keys() returns a view; list(...) shows the keys as a list
print(list(dct.values()))                # values() returns a view; convert it to a list for display
print(list(dct.items()))                 # items() returns a view of (key,value) pairs
print('cat' in dct)                      # Check if a dictionary has a given key; prints "True"
````


It is relatively straightforward to add new entries, modify existing entries, or delete existing entries.


````python

dct['fish'] = 'wet'       # Set an entry in a dictionary
print(dct['fish'])        # Prints value of key 'fish'
dct['dog'] = 'friendly'   # Change the value of key 'dog'
print(dct)
del dct['cat']            # Remove an element from a dictionary
print(dct)
dct[12] = 8               # keys and values in a dictionary can be of different types
print(dct)
````


Sometimes, we may wish to obtain a value for a specific key, but do not know if the element exists. Running the following line creates an error message:


````python

print(dct['monkey'])  # KeyError: 'monkey' because this key is not in the dictionary
````


To prevent the error, the `get` method can be used (enter `help("dict")` to learn about all the methods for `dict` type).


````python

print(dct.get('monkey', 'N/A'))  # Get an element with a default; prints "N/A"
print(dct.get('fish', 'N/A'))    # Get an element with a default; prints "wet"
````


Dictionaries are very interesting from the data structures perspective -- they are related to the idea of hash tables.



**Sets** (reference: https://docs.python.org/3/tutorial/datastructures.html#sets). A set is an unordered collection of distinct elements. It is defined using curly braces.


````python

animals = {'cat', 'dog'}
print(animals, type(animals))
print('cat' in animals)    # Check if an element is in a set; prints "True"
print('fish' in animals)   # prints "False"
animals.add('cat')         # Adding an element that is already in the set does nothing
print(animals)
animals.remove('cat')      # Remove an element from a set
print(animals)
````


# Control flow of Python

Similarly to other languages, Python allows control flow through conditionals (e.g., if-else) and loops (e.g., for loops, while loops).

## if-elif-else


````python

b = 2
if b < 0:
    print("It's negative")
elif b == 0:
    print('Equal to zero')
elif 0 < b < 5:
    print('Positive but smaller than 5')
else:
    print('Positive and larger than 5')
````


As you can see, the main difference is in syntax. `if-elif-else` does not require parentheses. Instead, indentation is used to identify blocks of code. JupyterLab helps with indentation while you type, but indentation is part of Python syntax and should be watched carefully.

## While loops


````python

x = 256
total = 0
while x > 0:
    if total > 500:
        break        # leaves the loop
    total += x
    x = x // 2
    print(x, total)
print('finally, the values are: ', x, total)
````


## For loops

`for` loops are somewhat different than in Java, C, or Matlab. Rather than giving the user the ability to define both the iteration step and halting condition (as in C), Python's `for` statement iterates over the items of an iterable object (for example, a list, string, dictionary, or range). You will learn to like `for` loops once you get used to them.


````python

total = 0
for i in range(5):  # range(5) represents the values 0, 1, 2, 3, 4
    total += i
print(total)        # reports the sum of the values visited by the loop
````


Here is another example that iterates through the elements of a provided list.


````python

sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:      # `is None` is the standard way to test for the special value None
        continue           # skip the rest of the block, keep looping
    total += value
print(total)
````


`for` loops can also iterate over dictionary items:


````python

dct = {'cat': 'cute', 'dog': 'furry', 'fish': 'wet'}  # Create a new dictionary with some data
for key in dct:                    # iterates over the keys of a dictionary
    print(key)
for key, value in dct.items():     # items() lets us iterate over (key,value) pairs
    print(key, 'is', value)
````


The role of keyword `in` might be confusing because it appears both in membership tests and in `for` loops. If you are curious about the distinction, the following Stack Overflow discussion may be helpful: https://stackoverflow.com/questions/38204342/python-in-keyword-in-expression-vs-in-for-loop



## try-except-else-finally

The `try-except-else-finally` control flow is useful for handling errors and exceptions in Python code. It can be useful in data science, for example when dealing with external files or converting imperfect data. You can read more at https://docs.python.org/3/tutorial/errors.html#handling-exceptions.


````python

try:
    number = int('not a number')           # this conversion will fail
except ValueError:
    print('The string could not be converted to an integer.')
````


## List Comprehensions

List comprehensions provide a concise way to generate lists and can come in handy.


````python

print([x*x for x in range(1,6)])    # prints a list of the squares of the first 5 positive integers
````

````python

strings = ['a', 'as', 'bat', 'car', 'dove', 'python']
print([x.upper() for x in strings if len(x) > 2])   # uppercase words longer than 2 characters
````

````python

print({val:index for index, val in enumerate(strings)})  # dictionary created from the strings list
````


# Functions

Python functions are defined using the `def` keyword.


````python

def sign(x):             # defining a function called sign
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 0         # a Python function can return different types depending on input

for x in [-1, 0, 1]:
    print(sign(x))
````


Functions are allowed to take optional keyword arguments.


````python

def hello(name, loud=False):      # the second input is optional at call time
    if loud:
        print(f'HELLO, {name.upper()}')
    else:
        print(f'Hello, {name}!')

hello('Bob')
hello('Fred', True)
````


There is much more to functions. A good starting point to learn more is the Python tutorial: https://docs.python.org/3/tutorial/controlflow.html#defining-functions.



# Classes

The syntax for defining classes in Python is not surprising. Doing good data science does not require writing many new classes, but understanding how they are created can be useful. In machine-learning code you will also frequently use objects created from classes provided by libraries.


````python

class Greeter:

    # Constructor
    def __init__(self, name):
        self.name = name  # Create an instance variable

    # Instance method
    def greet(self, loud=False):
        if loud:
            print(f'HELLO, {self.name.upper()}!')
        else:
            print(f'Hello, {self.name}')

g = Greeter('Fred')       # Construct an instance of the Greeter class
g.greet()                 # Call an instance method; prints "Hello, Fred"
g.greet(loud=True)        # Call an instance method; prints "HELLO, FRED!"
````


In Python, functions are "first-class". This is just a fancy way of saying that you can pass functions to other functions.


````python

def sum_of_anything(x, y, f):
    print(x, y, f)
    return f(x) + f(y)

sum_of_anything(3, 4, lambda x: x*x)    # lambda describes an anonymous square function
````


For those wanting to learn more about classes, a good starting point is https://docs.python.org/3/tutorial/classes.html.


## A short introduction to NumPy

NumPy is one of the most important packages for numerical work in Python and is used directly or indirectly by many machine-learning tools. A NumPy **array** is similar in spirit to a list, but it is designed for numerical computation. Arrays have a `shape`, usually contain values of a single data type, and support operations on many values at once.

Do not worry about learning NumPy in detail here. The goal is simply to recognize the object and a few common operations that will appear later in the course.


````python

x = np.array([1.0, 2.0, 3.0])   # create a one-dimensional NumPy array
print(x)                         # print the array
print(x.shape)                   # shape tells us the size of each dimension
print(x.dtype)                   # dtype tells us the type used to store the numbers

print(x * 2)                     # multiply every element by 2
print(x.mean())                  # compute the mean of all elements

X = np.array([[1.0, 2.0, 3.0],  # create a two-dimensional array
              [4.0, 5.0, 6.0]])

print(X)
print(X.shape)                   # 2 rows and 3 columns
print(X[:, 1])                   # select all rows from column with index 1
````


## A short introduction to pandas

The pandas package is widely used for tabular data. Its most common object is a **DataFrame**, which can be thought of as a table with named columns. We will use pandas only occasionally in this course, so this is only a first look.


````python

import pandas as pd                     # pd is the conventional alias for pandas

df = pd.DataFrame({                     # create a small table from a dictionary
    'student': ['Ada', 'Grace', 'Alan', 'Edsger'],
    'score':   [91, 95, 88, 93]
})

print(df)                               # display the whole table
print(df['score'])                      # select one column
print(df['score'].mean())               # compute the mean of the score column

high_scores = df[df['score'] >= 92]     # keep rows whose score is at least 92
print(high_scores)
````
